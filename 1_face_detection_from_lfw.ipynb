```python
import numpy as np
import dlib
from sklearn.datasets import fetch_lfw_people
from google.colab.patches import cv2_imshow  # Import cv2_imshow for displaying images
import cv2
```

```python
lfw_people = fetch_lfw_people(min_faces_per_person=100, resize=0.4)
```

```python
x = lfw_people.images
target_names = lfw_people.target_names
```

```python
 face_dlib = dlib.get_frontal_face_detector()
```

```python
def draw_bounding_boxes_dlib(image):
    # Convert image to uint8 format
    #image = (image * 255).astype(np.uint8)
    image = cv2.resize(image, (image.shape[1] * 3, image.shape[0] * 3))
    if len(image.shape) == 3 and image.shape[2] == 3:
        converted_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    else:
        converted_image = image
    faces = face_dlib(converted_image, 1)
    if len(faces) == 0:
        print("No faces detected.")
    else:
        print(f"Detected {len(faces)} face(s).")
    for face in faces:
        x, y, w, h = (face.left(), face.top(), face.width(), face.height())
        cv2.rectangle(converted_image, (x, y), (x + w, y + h), (255, 0, 0), 2)

    return converted_image
```

```python
print("dlib:")
for i in range(10):
    test_image = x[i]
    result_image = draw_bounding_boxes_dlib(test_image)
    cv2_imshow(result_image)
```
