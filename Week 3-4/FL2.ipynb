```python
import pickle
import cv2
import numpy as np
from PIL import Image as Img
from numpy import asarray, expand_dims
from keras_facenet import FaceNet


```

```python

# Initialize FaceNet and Haar Cascade
MyFaceNet = FaceNet()
HaarCascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Load the embeddings database
with open("data.pkl", "rb") as myfile:
    database = pickle.load(myfile)



```

```python

# Start video capture
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Detect multiple faces
    faces = HaarCascade.detectMultiScale(frame, scaleFactor=1.1, minNeighbors=5)
    
    for (x, y, w, h) in faces:
        # Extract face ROI
        face = frame[y:y+h, x:x+w]

        # Preprocess face (RGB conversion, resize, normalization)
        face_rgb = cv2.cvtColor(face, cv2.COLOR_BGR2RGB)
        face_pil = Img.fromarray(face_rgb)
        face_resized = face_pil.resize((160, 160))
        face_array = asarray(face_resized, dtype='float32')
        
        # Normalize face
        mean, std = face_array.mean(), face_array.std()
        face_normalized = (face_array - mean) / std
        
        # Get face embedding
        face_expanded = expand_dims(face_normalized, axis=0)
        signature = MyFaceNet.embeddings(face_expanded)
        
        # Identify the face
        min_dist = 0.9  # Dynamic threshold; experiment with this value
        identity = "Unknown"
        
        for name, db_embedding in database.items():
            dist = np.linalg.norm(db_embedding - signature)
            if dist < min_dist:
                min_dist = dist
                identity = name

        # Draw the face rectangle and label
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(frame, identity, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 0), 2)
    
    # Display the output
    cv2.imshow('Face Recognition', frame)

    # Exit on 'Esc' key press
    if cv2.waitKey(5) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
