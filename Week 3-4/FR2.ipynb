```python
import os
import pickle
import cv2
import numpy as np
from PIL import Image as Img
from numpy import asarray, expand_dims
from keras_facenet import FaceNet
```

```python
# Initialize FaceNet and Haar Cascade
MyFaceNet = FaceNet()
HaarCascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
folder = 'photo/'
database = {}
```

```python
for filename in os.listdir(folder):
    path = os.path.join(folder, filename)
    img = cv2.imread(path)
    
```

```python
    faces = HaarCascade.detectMultiScale(img, scaleFactor=1.1, minNeighbors=5)
```

```python

    
    # Store average embedding if multiple faces are detected
    face_embeddings = []
    for (x, y, w, h) in faces:
        # Extract the face
        face = img[y:y+h, x:x+w]
        
        # Preprocess face (convert to RGB, resize, normalize)
        face_rgb = cv2.cvtColor(face, cv2.COLOR_BGR2RGB)
        face_pil = Img.fromarray(face_rgb)
        face_resized = face_pil.resize((160, 160))
        face_array = asarray(face_resized, dtype='float32')
        
        # Normalize face
        mean, std = face_array.mean(), face_array.std()
        face_normalized = (face_array - mean) / std
        
        # Expand dimensions and get embedding
        face_expanded = expand_dims(face_normalized, axis=0)
        signature = MyFaceNet.embeddings(face_expanded)
        face_embeddings.append(signature)

    # If multiple embeddings, average them for a stable representation
    if face_embeddings:
        database[os.path.splitext(filename)[0]] = np.mean(face_embeddings, axis=0)


```

```python
# Save embeddings to file
with open("data.pkl", "wb") as myfile:
    pickle.dump(database, myfile)
```
